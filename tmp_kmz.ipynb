```python
photo_folder = 'C:/Users/au646858/Downloads/Photos-1-001/'
photos = [photo_folder+x for x in os.listdir(photo_folder) if x.endswith('.jpg')]

import exifread
import os

def print_detailed_metadata(image_path):
    """Print detailed metadata with descriptions"""
    print(f"\n{'='*80}")
    print(f"DETAILED METADATA for: {os.path.basename(image_path)}")
    print(f"{'='*80}")
    
    try:
        with open(image_path, 'rb') as f:
            tags = exifread.process_file(f, details=True)
        
        if not tags:
            print("No EXIF data found")
            return
        
        # Print file info
        print(f"File size: {os.path.getsize(image_path):,} bytes")
        
        # Sort and print all tags
        for tag in sorted(tags.keys()):
            try:
                value = tags[tag]
                print(f"{tag:30}: {value}")
                
                # For some tags, show additional details
                if hasattr(value, 'values') and tag.startswith('GPS'):
                    if value.values:
                        print(f"{' '*30}  Raw values: {value.values}")
                        
            except Exception as e:
                print(f"{tag:30}: Error reading value - {e}")
    
    except Exception as e:
        print(f"Error reading {image_path}: {e}")

# Print detailed metadata for all photos
for photo in photos:
    print_detailed_metadata(photo)
```

```python
import exifread
import os
import shutil

def convert_to_degrees(value):
    """Convert GPS coordinates from DMS format to decimal degrees"""
    d = float(value.values[0].num) / float(value.values[0].den)
    m = float(value.values[1].num) / float(value.values[1].den)
    s = float(value.values[2].num) / float(value.values[2].den)
    return d + (m / 60.0) + (s / 3600.0)

def get_gps_coordinates(image_path):
    """Extract GPS coordinates from image EXIF data"""
    try:
        with open(image_path, 'rb') as f:
            tags = exifread.process_file(f)
        
        lat = None
        lon = None
        
        # Check if GPS data exists
        if 'GPS GPSLatitude' in tags and 'GPS GPSLatitudeRef' in tags:
            lat = convert_to_degrees(tags['GPS GPSLatitude'])
            if str(tags['GPS GPSLatitudeRef']) == 'S':
                lat = -lat
        
        if 'GPS GPSLongitude' in tags and 'GPS GPSLongitudeRef' in tags:
            lon = convert_to_degrees(tags['GPS GPSLongitude'])
            if str(tags['GPS GPSLongitudeRef']) == 'W':
                lon = -lon
        
        return lat, lon
    
    except Exception as e:
        print(f"Error reading {image_path}: {e}")
        return None, None

# Test with your photos
photo_folder = 'C:/Users/au646858/Downloads/Photos-1-001/'
new_folder = 'C:/Users/au646858/Documents/vaults/fieldwork/files/'
photos = [photo_folder + x for x in os.listdir(photo_folder) if x.endswith('.jpg')]

print("GPS Coordinates for each photo:")
print("-" * 60)

# Create markdown file with GPS coordinates
with open(new_folder + 'dollow_bank_height2.md', 'w') as f:
    f.write('# Dollow bank height estimates\n\n')
    
    for photo in photos:
        lat, lon = get_gps_coordinates(photo)
        photo_name = os.path.split(photo)[-1]
        print(photo_name, lat, lon)
```

```python
bank_heights = {}
with open(new_folder + 'dollow_bank_height_table.md', 'w') as fout:
    fout.write("| Photo Name | Latitude | Longitude | Bank Height |\n")
    fout.write("|------------|----------|-----------|-------------|\n")

    with open(new_folder + 'dollow_bank_height.md', 'r') as f:
        lines = f.readlines()
        for line in lines:
            line = line.strip()
            if line.startswith('!['):
                parts = line.split('|')
                if len(parts) > 1:
                    photo_name = parts[0][3:]
                    print(f"Found photo: {photo_name}")
                    lat,lon = get_gps_coordinates(photo_folder + photo_name)
            if line.startswith('bank height:'):
                parts = line.split(':')
                if len(parts) > 1:
                    bank_height = parts[1].strip()
                    print(f"Bank height: {bank_height}")
                    fout.write(f"| {photo_name} | {lat} | {lon} | {bank_height} |\n")
                    bank_heights[photo_name] = bank_height
```

```python
def create_kml_with_photos(output_file):
    """Create KML file with photos at their GPS locations"""
    

    kml_content = '''<?xml version="1.0" encoding="UTF-8"?>
<kml xmlns="http://www.opengis.net/kml/2.2">
  <Document>
    <name>Dollow Bank Height Survey</name>
    <description>River bank height measurements with photos</description>
    
    <Style id="photoIcon">
      <IconStyle>
        <Icon>
          <href>http://maps.google.com/mapfiles/kml/shapes/camera.png</href>
          <scale>1.2</scale>
        </Icon>
      </IconStyle>
    </Style>
'''
    
    for photo in photos:
        lat, lon = get_gps_coordinates(photo)
        photo_name = os.path.split(photo)[-1]
        
        if lat is not None and lon is not None:
            # Get bank height or use placeholder
            bank_height = bank_heights.get(photo_name, 'Not measured')
            
            # Convert photo to base64 for embedding (optional)
            # For web hosting, you might want to upload photos and use URLs instead
            
            kml_content += f'''
    <Placemark>
      <name>{photo_name}</name>
      <description><![CDATA[
        <h3>Bank Height: {bank_height}</h3>
        <img src="{photo}" width="400" style="max-width:400px;"/>
        <br/>
        <p>Coordinates: {lat:.6f}, {lon:.6f}</p>
        <p>Photo: {photo_name}</p>
      ]]></description>
      <styleUrl>#photoIcon</styleUrl>
      <Point>
        <coordinates>{lon},{lat},0</coordinates>
      </Point>
    </Placemark>'''
    
    kml_content += '''
  </Document>
</kml>'''
    
    with open(output_file, 'w', encoding='utf-8') as f:
        f.write(kml_content)
    
    print(f"KML file created: {output_file}")

# Create the KML file
kml_output = new_folder + 'dollow_bank_survey.kml'
create_kml_with_photos(kml_output)
```

```python
import zipfile
import os

def create_kmz_with_embedded_photos(output_file):
    """Create KMZ file with embedded photos - fixed for Google Earth"""
    
    # Create KML content with proper relative paths
    kml_content = '''<?xml version="1.0" encoding="UTF-8"?>
<kml xmlns="http://www.opengis.net/kml/2.2">
  <Document>
    <name>Dollow Bank Height Survey</name>
    <description>River bank height measurements with photos</description>
    
    <Style id="photoIcon">
      <IconStyle>
        <Icon>
          <href>http://maps.google.com/mapfiles/kml/shapes/camera.png</href>
          <scale>1.5</scale>
        </Icon>
        <hotSpot x="0.5" y="0" xunits="fraction" yunits="fraction"/>
      </IconStyle>
      <BalloonStyle>
        <bgColor>ffffffff</bgColor>
        <textColor>ff000000</textColor>
        <text>$[description]</text>
      </BalloonStyle>
    </Style>
'''
    
    # Create KMZ file (ZIP archive)
    with zipfile.ZipFile(output_file, 'w', zipfile.ZIP_DEFLATED) as kmz:
        
        for photo in photos:
            lat, lon = get_gps_coordinates(photo)
            photo_name = os.path.split(photo)[-1]
            
            if lat is not None and lon is not None:
                # Get bank height or use placeholder
                bank_height = bank_heights.get(photo_name, 'Not measured')
                
                # Add photo directly to root of KMZ (not in subfolder)
                kmz.write(photo, photo_name)
                
                # Reference photo with just filename (no folder path)
                kml_content += f'''
    <Placemark>
      <name>Bank Height: {bank_height}</name>
      <description><![CDATA[
        <div style="font-family: Arial, sans-serif;">
          <h3 style="color: #2E8B57;">Bank Height: {bank_height}</h3>
          <img src="{photo_name}" width="400" style="max-width:400px; border: 2px solid #ddd; border-radius: 5px;"/>
          <br/><br/>
          <table style="border-collapse: collapse; width: 100%;">
            <tr><td style="font-weight: bold;">Location:</td><td>{lat:.6f}°N, {lon:.6f}°E</td></tr>
            <tr><td style="font-weight: bold;">Photo:</td><td>{photo_name}</td></tr>
          </table>
        </div>
      ]]></description>
      <styleUrl>#photoIcon</styleUrl>
      <Point>
        <coordinates>{lon},{lat},0</coordinates>
      </Point>
    </Placemark>'''
        
        kml_content += '''
  </Document>
</kml>'''
        
        # Add KML file to KMZ root
        kmz.writestr('doc.kml', kml_content.encode('utf-8'))
    
    print(f"KMZ file created: {output_file}")
    print(f"File size: {os.path.getsize(output_file):,} bytes")

# Create the KMZ file
kmz_output = new_folder + 'dollow_bank_survey.kmz'
create_kmz_with_embedded_photos(kmz_output)
```
