```python
from plogly import Log
import numpy as np
```

```python
logrho = np.array([20, 80, 40])
lg = Log(logrho, [0, 10-1, 20-1], [10-1, 20-1, 40], '')
fig = lg.plot(cmap='turbo', x_offset=-1, zmin=0, zmax=100)

for i in range(15):
    lg = Log(logrho, [0, 10+i, 20+i], [10+i, 20+i, 40], '')
    lg.plot(cmap='turbo', x_offset=i, fig=fig,  zmin=0, zmax=100)
fig
```
